```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
customers = pd.read_csv("olist_customers_dataset.csv")
geolocation = pd.read_csv("olist_geolocation_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
category_translation = pd.read_csv("product_category_name_translation.csv")

print("All 9 datasets loaded successfully!")
```

```python
datasets = {
    "Customers": customers,
    "Geolocation": geolocation,
    "Order Items": order_items,
    "Payments": payments,
    "Reviews": reviews,
    "Orders": orders,
    "Products": products,
    "Sellers": sellers,
    "Category Translation": category_translation
}

inventory = []

for name, df in datasets.items():
    inventory.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

inventory_df = pd.DataFrame(inventory)

display(inventory_df)
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    errors="coerce"
)

print("Earliest Order:",
      orders["order_purchase_timestamp"].min())

print("Latest Order:",
      orders["order_purchase_timestamp"].max())

print("\nOrder Status Distribution:")
display(orders["order_status"].value_counts())
```

```python
for name, df in datasets.items():

    missing_count = df.isnull().sum()
    missing_pct = (missing_count / len(df)) * 100

    missing_summary = pd.DataFrame({
        "Missing Count": missing_count,
        "Missing %": missing_pct.round(2)
    })

    missing_summary = missing_summary[
        missing_summary["Missing Count"] > 0
    ]

    if not missing_summary.empty:
        print(f"\n{name}")
        display(missing_summary)
```

```python
# Count missing delivery dates by order status

delivery_check = orders.groupby(
    "order_status"
).agg(
    total_orders=("order_id", "count"),
    missing_carrier_dates=(
        "order_delivered_carrier_date",
        lambda x: x.isna().sum()
    ),
    missing_customer_dates=(
        "order_delivered_customer_date",
        lambda x: x.isna().sum()
    )
)

display(delivery_check)
```

```python
# Phase 1, Step 2: Key and relationship validation

key_checks = {
    "Duplicate order IDs":
        orders["order_id"].duplicated().sum(),

    "Duplicate customer IDs":
        customers["customer_id"].duplicated().sum(),

    "Duplicate product IDs":
        products["product_id"].duplicated().sum(),

    "Duplicate seller IDs":
        sellers["seller_id"].duplicated().sum(),

    "Duplicate order-item keys":
        order_items.duplicated(
            ["order_id", "order_item_id"]
        ).sum(),

    "Orders without matching customers":
        (~orders["customer_id"].isin(
            customers["customer_id"]
        )).sum(),

    "Items without matching orders":
        (~order_items["order_id"].isin(
            orders["order_id"]
        )).sum(),

    "Items without matching products":
        (~order_items["product_id"].isin(
            products["product_id"]
        )).sum(),

    "Items without matching sellers":
        (~order_items["seller_id"].isin(
            sellers["seller_id"]
        )).sum(),

    "Payments without matching orders":
        (~payments["order_id"].isin(
            orders["order_id"]
        )).sum(),

    "Reviews without matching orders":
        (~reviews["order_id"].isin(
            orders["order_id"]
        )).sum()
}

key_validation_df = pd.DataFrame(
    list(key_checks.items()),
    columns=["Validation Check", "Violations"]
)

display(key_validation_df)
```

```python
# Convert order purchase dates

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    errors="coerce"
)

print(
    "Earliest purchase:",
    orders["order_purchase_timestamp"].min()
)

print(
    "Latest purchase:",
    orders["order_purchase_timestamp"].max()
)

print(
    "Invalid purchase dates:",
    orders["order_purchase_timestamp"].isna().sum()
)

# Number of orders by year

orders["purchase_year"] = (
    orders["order_purchase_timestamp"].dt.year
)

display(
    orders["purchase_year"]
    .value_counts()
    .sort_index()
)
```

```python
key_validation_df.to_csv(
    "key_validation_results.csv",
    index=False
)

print("Validation report saved successfully!")
```

```python
dictionary_rows = []

for table_name, df in datasets.items():
    for column in df.columns:
        dictionary_rows.append({
            "Table": table_name,
            "Column": column,
            "Python_Type": str(df[column].dtype),
            "Non_Null_Count": int(df[column].count()),
            "Missing_Count": int(df[column].isna().sum()),
            "Missing_Percent": round(
                df[column].isna().mean() * 100, 2
            ),
            "Unique_Values": int(df[column].nunique()),
            "MySQL_Type": "",
            "Key_Type": "",
            "Business_Definition": "",
            "Cleaning_Rule": ""
        })

column_dictionary = pd.DataFrame(dictionary_rows)

display(column_dictionary.head(15))

column_dictionary.to_excel(
    "olist_column_dictionary.xlsx",
    index=False
)

print("Column dictionary exported!")
```
