```python
customers["customer_zip_code_prefix"] = (
    customers["customer_zip_code_prefix"]
    .astype("string")
    .str.zfill(5)
)

geolocation["geolocation_zip_code_prefix"] = (
    geolocation["geolocation_zip_code_prefix"]
    .astype("string")
    .str.zfill(5)
)

sellers["seller_zip_code_prefix"] = (
    sellers["seller_zip_code_prefix"]
    .astype("string")
    .str.zfill(5)
)

print("ZIP prefixes converted to five-character strings.")
```

```python
# Preserve ZIP prefixes as strings
for df, col in [
    (customers, "customer_zip_code_prefix"),
    (sellers, "seller_zip_code_prefix"),
    (geolocation, "geolocation_zip_code_prefix")
]:
    df[col] = df[col].astype("string").str.zfill(5)

# Define key classifications
primary_keys = {
    "Customers": ["customer_id"],
    "Orders": ["order_id"],
    "Products": ["product_id"],
    "Sellers": ["seller_id"],
    "Order Items": ["order_id", "order_item_id"],
    "Payments": ["order_id", "payment_sequential"],
    "Reviews": ["review_id", "order_id"],
    "Category Translation": ["product_category_name"]
}

foreign_keys = {
    "Orders": ["customer_id"],
    "Order Items": ["order_id", "product_id", "seller_id"],
    "Payments": ["order_id"],
    "Reviews": ["order_id"],
    "Products": ["product_category_name"]
}

def mysql_type(column, dtype):
    if "zip_code_prefix" in column:
        return "CHAR(5)"
    if column.endswith("_id") and column != "order_item_id":
        return "VARCHAR(32)"
    if "date" in column or "timestamp" in column or column == "order_approved_at":
        return "DATETIME"
    if column in ["price", "freight_value", "payment_value"]:
        return "DECIMAL(12,2)"
    if column in ["geolocation_lat", "geolocation_lng"]:
        return "DECIMAL(11,8)"
    if "comment" in column:
        return "TEXT"
    if "int" in str(dtype):
        return "INT"
    if "float" in str(dtype):
        return "DECIMAL(12,2)"
    return "VARCHAR(100)"

rows = []

for table, df in datasets.items():
    for column in df.columns:
        is_pk = column in primary_keys.get(table, [])
        is_fk = column in foreign_keys.get(table, [])

        key_type = (
            "PK/FK" if is_pk and is_fk else
            "PK" if is_pk else
            "FK" if is_fk else "None"
        )

        rows.append({
            "Table": table,
            "Column": column,
            "Python_Type": str(df[column].dtype),
            "MySQL_Type": mysql_type(column, df[column].dtype),
            "Key_Type": key_type,
            "Missing_Count": int(df[column].isna().sum()),
            "Unique_Values": int(df[column].nunique()),
            "Business_Definition": "",
            "Cleaning_Rule": ""
        })

final_dictionary = pd.DataFrame(rows)

display(final_dictionary.head(20))

final_dictionary.to_excel(
    "olist_data_dictionary.xlsx",
    index=False
)

print("Excel dictionary created successfully!")
```

```python
import pandas as pd
import numpy as np

# Creating copies to preserve original Dataset
cleaned = {
    name: df.copy(deep=True)
    for name, df in datasets.items()
}

print("Working copies created:")
for name, df in cleaned.items():
    print(name, df.shape)
```

```python
# Standardizing Zip-code prefixes
zip_columns = {
    "Customers": "customer_zip_code_prefix",
    "Sellers": "seller_zip_code_prefix",
    "Geolocation": "geolocation_zip_code_prefix"
}

for table, column in zip_columns.items():
    cleaned[table][column] = (
        cleaned[table][column]
        .astype("string")
        .str.replace(r"\.0$", "", regex=True)
        .str.zfill(5)
    )

    print(
        table,
        cleaned[table][column].head().tolist()
    )
```

```python
# Converting timestamps
date_columns = {
    "Orders": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],
    "Order Items": [
        "shipping_limit_date"
    ],
    "Reviews": [
        "review_creation_date",
        "review_answer_timestamp"
    ]
}

for table, columns in date_columns.items():
    for column in columns:
        cleaned[table][column] = pd.to_datetime(
            cleaned[table][column],
            errors="coerce"
        )

print("Timestamp conversion completed.")
```

```python
# Comparing null counts before and after
date_audit = []

for table, columns in date_columns.items():
    for column in columns:
        before = datasets[table][column].isna().sum()
        after = cleaned[table][column].isna().sum()

        date_audit.append({
            "Table": table,
            "Column": column,
            "Original_Nulls": before,
            "Cleaned_Nulls": after,
            "New_Invalid_Dates": after - before
        })

date_audit_df = pd.DataFrame(date_audit)
display(date_audit_df)
```

```python
# Validating numeric values
numeric_issues = {}

# Financial fields
for table, columns in {
    "Order Items": ["price", "freight_value"],
    "Payments": ["payment_value"]
}.items():
    for column in columns:
        values = pd.to_numeric(
            cleaned[table][column],
            errors="coerce"
        )

        numeric_issues[f"{table}.{column}"] = {
            "Negative": int((values < 0).sum()),
            "Missing_or_Invalid": int(values.isna().sum())
        }

# Review scores
scores = pd.to_numeric(
    cleaned["Reviews"]["review_score"],
    errors="coerce"
)

numeric_issues["Reviews.review_score"] = {
    "Outside_1_to_5": int(
        (~scores.between(1, 5)).sum()
    )
}

# Product dimensions
dimension_cols = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for column in dimension_cols:
    values = pd.to_numeric(
        cleaned["Products"][column],
        errors="coerce"
    )

    numeric_issues[f"Products.{column}"] = {
        "Negative": int((values < 0).sum()),
        "Missing_or_Invalid": int(values.isna().sum())
    }

display(pd.DataFrame(numeric_issues).T)
```

```python
# Validating order timestamps
orders_clean = cleaned["Orders"]

# Delivered before purchase
delivery_before_purchase = (
    orders_clean["order_delivered_customer_date"].notna()
    & (
        orders_clean["order_delivered_customer_date"]
        < orders_clean["order_purchase_timestamp"]
    )
)

# Carrier handoff before purchase
carrier_before_purchase = (
    orders_clean["order_delivered_carrier_date"].notna()
    & (
        orders_clean["order_delivered_carrier_date"]
        < orders_clean["order_purchase_timestamp"]
    )
)

# Delivered status without actual delivery date
delivered_missing_date = (
    orders_clean["order_status"].eq("delivered")
    & orders_clean["order_delivered_customer_date"].isna()
)

print("Delivery before purchase:",
      delivery_before_purchase.sum())

print("Carrier handoff before purchase:",
      carrier_before_purchase.sum())

print("Delivered orders missing delivery date:",
      delivered_missing_date.sum())
```

```python
# Creating ZIP-level geographic lookup

geo = cleaned["Geolocation"].copy()

# Keep plausible Brazilian coordinates
geo_valid = geo[
    geo["geolocation_lat"].between(-35, 6)
    & geo["geolocation_lng"].between(-75, -30)
].copy()

# Remove exact duplicates in the derived copy
geo_valid = geo_valid.drop_duplicates()

# Calculate representative coordinates per ZIP
geo_lookup = (
    geo_valid
    .groupby("geolocation_zip_code_prefix")
    .agg(
        latitude=("geolocation_lat", "median"),
        longitude=("geolocation_lng", "median"),
        observation_count=("geolocation_lat", "size")
    )
    .reset_index()
)

print("Geographic lookup rows:", len(geo_lookup))
print(
    "Unique ZIP prefixes:",
    geo_lookup["geolocation_zip_code_prefix"].nunique()
)

print(
    "Duplicate lookup ZIP prefixes:",
    geo_lookup["geolocation_zip_code_prefix"].duplicated().sum()
)

display(geo_lookup.head(10))
```

```python
# Final relational integrity validation
c = cleaned

checks = {
    "Duplicate customer IDs":
        c["Customers"]["customer_id"].duplicated().sum(),

    "Duplicate order IDs":
        c["Orders"]["order_id"].duplicated().sum(),

    "Duplicate product IDs":
        c["Products"]["product_id"].duplicated().sum(),

    "Duplicate seller IDs":
        c["Sellers"]["seller_id"].duplicated().sum(),

    "Duplicate order-item keys":
        c["Order Items"].duplicated(
            ["order_id", "order_item_id"]
        ).sum(),

    "Duplicate payment keys":
        c["Payments"].duplicated(
            ["order_id", "payment_sequential"]
        ).sum(),

    "Duplicate review-order keys":
        c["Reviews"].duplicated(
            ["review_id", "order_id"]
        ).sum(),

    "Orders without customers":
        (~c["Orders"]["customer_id"].isin(
            c["Customers"]["customer_id"]
        )).sum(),

    "Items without orders":
        (~c["Order Items"]["order_id"].isin(
            c["Orders"]["order_id"]
        )).sum(),

    "Items without products":
        (~c["Order Items"]["product_id"].isin(
            c["Products"]["product_id"]
        )).sum(),

    "Items without sellers":
        (~c["Order Items"]["seller_id"].isin(
            c["Sellers"]["seller_id"]
        )).sum(),

    "Payments without orders":
        (~c["Payments"]["order_id"].isin(
            c["Orders"]["order_id"]
        )).sum(),

    "Reviews without orders":
        (~c["Reviews"]["order_id"].isin(
            c["Orders"]["order_id"]
        )).sum()
}

integrity_report = pd.DataFrame(
    checks.items(),
    columns=["Validation_Check", "Violations"]
)

display(integrity_report)

print(
    "All integrity checks passed:",
    integrity_report["Violations"].eq(0).all()
)
```

```python
# Exporting clean CSV files

output_names = {
    "Customers": "clean_customers.csv",
    "Geolocation": "clean_geolocation_staging.csv",
    "Order Items": "clean_order_items.csv",
    "Payments": "clean_order_payments.csv",
    "Reviews": "clean_order_reviews.csv",
    "Orders": "clean_orders.csv",
    "Products": "clean_products.csv",
    "Sellers": "clean_sellers.csv",
    "Category Translation": "clean_category_translation.csv"
}

# Stop if any relational validation failed
assert integrity_report["Violations"].eq(0).all(), (
    "Relational integrity validation failed."
)

# Ensure exported schemas contain expected columns
assert schema_check_df["Match"].all(), (
    "Column-count validation failed."
)

assert reconciliation_df["Difference"].eq(0).all(), (
    "Row-count reconciliation failed."
)

for table, filename in output_names.items():
    cleaned[table].to_csv(
        filename,
        index=False,
        encoding="utf-8",
        date_format="%Y-%m-%d %H:%M:%S"
    )

    print(f"Exported: {filename}")

# Export derived geographic lookup separately
geo_lookup.to_csv(
    "geo_zip_lookup.csv",
    index=False,
    encoding="utf-8"
)

print("All CSV exports completed.")
```
